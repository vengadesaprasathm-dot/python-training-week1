# Day 5 – Task 3: Five Visualisations

This notebook creates five charts from the employee dataset used in Day 5 Task 2.
Every figure is saved automatically in `reports/figures/`.


## 1. Import libraries and load the dataset

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Run this notebook from the Day5_Task3_Five_Visualisations folder.
# The CSV is reused from the Task 2 folder.
data_path = Path("../Day5_Task2_Exploratory_Analysis/employee_data.csv")
df = pd.read_csv(data_path)

figures_dir = Path("reports/figures")
figures_dir.mkdir(parents=True, exist_ok=True)

# Consistent chart style and colour scheme
sns.set_theme(style="whitegrid")
PALETTE = "Set2"
MAIN_COLOR = "#4472C4"

print("Dataset shape:", df.shape)
display(df.head())
```

## 2. Prepare the columns

```python
# Check that the columns needed for these charts are available.
required_columns = ["Department", "Age", "Monthly_Income"]
missing_columns = [column for column in required_columns if column not in df.columns]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Monthly_Income"] = pd.to_numeric(df["Monthly_Income"], errors="coerce")
chart_df = df.dropna(subset=required_columns).copy()

print("Rows available for charts:", len(chart_df))
```

## 3. Bar chart – Average monthly income by department

```python
department_income = (
    chart_df.groupby("Department", as_index=False)["Monthly_Income"]
    .mean()
    .sort_values("Monthly_Income", ascending=False)
)

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=department_income, x="Department", y="Monthly_Income",
            hue="Department", palette=PALETTE, legend=False, ax=ax)
ax.set_title("Average Monthly Income by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Average monthly income (currency units)")
ax.tick_params(axis="x", rotation=25)
fig.tight_layout()
fig.savefig(figures_dir / "01_bar_average_income_by_department.png", dpi=300, bbox_inches="tight")
plt.show()

print("Insight: The department with the tallest bar has the highest average monthly income.")
```

## 4. Histogram – Employee age distribution

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.histplot(data=chart_df, x="Age", bins=10, color=MAIN_COLOR,
             edgecolor="white", ax=ax)
ax.set_title("Distribution of Employee Ages")
ax.set_xlabel("Age (years)")
ax.set_ylabel("Number of employees (employees)")
fig.tight_layout()
fig.savefig(figures_dir / "02_histogram_employee_age.png", dpi=300, bbox_inches="tight")
plt.show()

print("Insight: The tallest age interval shows the age range containing the most employees.")
```

## 5. Scatter plot – Age versus monthly income

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(data=chart_df, x="Age", y="Monthly_Income",
                hue="Department", palette=PALETTE, s=75, ax=ax)
ax.set_title("Employee Age vs Monthly Income")
ax.set_xlabel("Age (years)")
ax.set_ylabel("Monthly income (currency units)")
ax.legend(title="Department", bbox_to_anchor=(1.02, 1), loc="upper left")
fig.tight_layout()
fig.savefig(figures_dir / "03_scatter_age_vs_income.png", dpi=300, bbox_inches="tight")
plt.show()

print("Insight: The point pattern indicates whether monthly income tends to increase, decrease, or show no clear relationship with age.")
```

## 6. Line chart – Average monthly income by age

```python
age_income = (
    chart_df.groupby("Age", as_index=False)["Monthly_Income"]
    .mean()
    .sort_values("Age")
)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(age_income["Age"], age_income["Monthly_Income"],
        marker="o", linewidth=2, color=MAIN_COLOR, label="Average monthly income")
ax.set_title("Average Monthly Income by Age")
ax.set_xlabel("Age (years)")
ax.set_ylabel("Average monthly income (currency units)")
ax.legend()
fig.tight_layout()
fig.savefig(figures_dir / "04_line_average_income_by_age.png", dpi=300, bbox_inches="tight")
plt.show()

print("Insight: This sorted line chart shows how average monthly income changes across employee ages.")
```

## 7. Box plot – Monthly income by department

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=chart_df, x="Department", y="Monthly_Income",
            hue="Department", palette=PALETTE, legend=False, ax=ax)
ax.set_title("Monthly Income Distribution by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Monthly income (currency units)")
ax.tick_params(axis="x", rotation=25)
fig.tight_layout()
fig.savefig(figures_dir / "05_boxplot_income_by_department.png", dpi=300, bbox_inches="tight")
plt.show()

print("Insight: The box plot compares the median, spread, and possible outliers in monthly income across departments.")
```

## 8. Confirm saved figures

The five PNG files should be available in `reports/figures/`.


```python
saved_figures = sorted(figures_dir.glob("*.png"))
for figure in saved_figures:
    print(figure)

print(f"\nTotal figures saved: {len(saved_figures)}")
assert len(saved_figures) == 5, "Expected exactly five saved chart images."
```
